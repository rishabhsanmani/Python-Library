```python
import time
import numpy as np

size = 10000000
#for list
a = list(range(size))
start_l = time.time()
a = list(map(lambda i : i**2, a))
end_l = time.time() - start_l

#array
a = np.arange(size)
start_a = time.time()
a = [a**2]
end_a = time.time() - start_a

print("Time take by list:",end_l)
print("Time take by array",end_a)
```
